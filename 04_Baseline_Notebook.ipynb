# Customer Churn: Exploratory Non-ML Baseline
This notebook is a classroom demonstration using the supplied 12-row CSV. It is **not suitable for deployment**.

```python
import pandas as pd
from sklearn.metrics import confusion_matrix, classification_report

df = pd.read_csv('customer-churn-training.csv')
print('Shape:', df.shape)
display(df.head())
print('\nTarget counts:\n', df['churned'].value_counts())
print('\nMissing values:\n', df.isna().sum())
```

## Baseline A: majority class
Predict the most frequent class for every row. This is a simple reference point, not a useful customer-level intervention.

```python
y = df['churned']
majority_class = int(y.mode().iloc[0])
pred_majority = [majority_class] * len(y)
print('Majority class:', majority_class)
print(confusion_matrix(y, pred_majority, labels=[0, 1]))
print(classification_report(y, pred_majority, zero_division=0))
```

## Baseline B: transparent triage rule
Flag a customer for human review if `last_login_days >= 14` OR `support_tickets >= 4`. This rule is illustrative and should be reviewed with domain experts.

```python
pred_rule = ((df['last_login_days'] >= 14) | (df['support_tickets'] >= 4)).astype(int)
print('Flagged rows:', int(pred_rule.sum()), 'of', len(df))
print(confusion_matrix(y, pred_rule, labels=[0, 1]))
print(classification_report(y, pred_rule, zero_division=0))
display(df.assign(rule_flag=pred_rule))
```

## Interpretation and limitations
- The file has only 12 observations (5 churned, 7 not churned); metrics are highly unstable.
- No timestamps are supplied, so a chronological train/test split and leakage audit cannot be performed.
- The churn definition and prediction horizon are unknown.
- Do not train or deploy a machine-learning model from this sample. Use this notebook only to demonstrate a transparent baseline and data checks.
- Before future modelling, obtain a much larger time-stamped dataset, define the target horizon, exclude identifiers, audit leakage, and evaluate on a later time period.
